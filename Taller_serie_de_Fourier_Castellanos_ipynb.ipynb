```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
b = 1.0  # amplitud constante
a_values = [0.1, 0.5, 1.0]  # distintos valores del parámetro 'a'

# Rango temporal
t = np.linspace(-1, 1, 1000)
u = np.heaviside(t, 1)  # función escalón unitario u(t)

# Crear la figura
plt.figure(figsize=(8, 5))

# Graficar para distintos valores de 'a'
for a in a_values:
    x = b * np.exp(-10 * a * t) * u
    plt.plot(t, x, label=f'a = {a}')

# Personalización del gráfico
plt.title(r'Señal $x(t) = b\,e^{-10 a t}u(t)$ para distintos valores de $a$')
plt.xlabel('Tiempo t [s]')
plt.ylabel('Amplitud x(t)')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la señal
a = 2.0         # amplitud
omega = 2 * np.pi * 1  # frecuencia angular (1 Hz)

# Rango temporal
t = np.linspace(0, 2, 1000)  # 2 segundos, 1000 muestras

# Definición de las señales
x = a * np.sin(omega * t)
x2 = a**2 * (np.sin(omega * t))**2

# Gráficas
plt.figure(figsize=(10, 5))

plt.plot(t, x, label=r'$x(t) = a \sin(\omega t)$', linewidth=2)
plt.plot(t, x2, label=r'$x^2(t) = a^2 \sin^2(\omega t)$', linestyle='--', linewidth=2)

# Personalización
plt.title('Señal sinusoidal y su cuadrado')
plt.xlabel('Tiempo t [s]')
plt.ylabel('Amplitud')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Mostrar
plt.show()
```
