#*Entrega 3: Diseño de compensador, Sistemas Dinamicos y Control*

*Codigo de las respuestas del sistema a partir de la función de transferencia.*



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Parametros
L = 660e-6
C = 20e-6
R = 6.0
Vi = 12.0
D = 0.5

# Coeficientes
a2 = L * C
a1 = L / R
a0 = 1.0

# Ganancia efectiva (Vi * (1 - D))
num = [Vi * (1 - D)]
den = [a2, a1, a0]

system = signal.TransferFunction(num, den)

# Diagrama de Bode
w, mag, phase = signal.bode(system)
plt.figure(figsize=(8,6))
plt.subplot(2,1,1)
plt.semilogx(w, mag)
plt.title('Diagrama de Bode')
plt.ylabel('Magnitud (dB)')
plt.grid(True, which='both', ls=':')
plt.subplot(2,1,1)
plt.semilogx(w, mag, color='red')

# Fase
plt.subplot(2,1,2)
plt.semilogx(w, phase)
plt.ylabel('Fase (°)')
plt.xlabel('Frecuencia (rad/s)')
plt.grid(True, which='both', ls=':')
plt.tight_layout()
plt.subplot(2,1,2)
plt.semilogx(w, phase, color='red')
plt.show()

# Respuesta al impulso
t_imp, y_imp = signal.impulse(system)
plt.figure()
plt.plot(t_imp, y_imp)
plt.title('Respuesta al impulso')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.grid(True)
plt.plot(t_imp, y_imp, color='red')
plt.show()

# Respuesta al escalón
t_step, y_step = signal.step(system)
plt.figure()
plt.plot(t_step, y_step)
plt.title('Respuesta al escalón')
plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud')
plt.grid(True)
plt.plot(t_step, y_step, color='red')
plt.show()
```
