```python
# --- Gráfica de la señal periódica x(t) ---

import numpy as np
import matplotlib.pyplot as plt

# Parámetros
w0 = 2 * np.pi   # frecuencia fundamental (rad/s)
t = np.linspace(-2, 2, 1000)  # rango de tiempo

# Definición de la señal
x = (2/3)*np.cos(w0*t) + (1/5)*np.cos(5*w0*t) - (1/7)*np.cos(7*w0*t) + (1/9)*np.cos(9*w0*t)

# Gráfica
plt.figure(figsize=(10, 5))
plt.plot(t, x, 'b', linewidth=2, label=r"$x(t)=\frac{2}{3}\cos(\omega_0 t)+\frac{1}{5}\cos(5\omega_0 t)-\frac{1}{7}\cos(7\omega_0 t)+\frac{1}{9}\cos(9\omega_0 t)$")
plt.title("Gráfica de la señal periódica x(t)", fontsize=14)
plt.xlabel("Tiempo t [s]", fontsize=12)
plt.ylabel("Amplitud x(t)", fontsize=12)
plt.grid(True)
plt.legend()
plt.show()
```

```python
# --- Señal con más armónicos impares (aproximación a onda cuadrada) ---
import numpy as np
import matplotlib.pyplot as plt

w0 = 2 * np.pi
t = np.linspace(-2, 2, 2000)

# Serie de Fourier truncada con armónicos impares
x = np.zeros_like(t)
for n in range(1, 20, 2):  # 1, 3, 5, ..., 19
    signo = (-1)**((n-1)//2)
    x += signo * (1/n) * np.cos(n * w0 * t)

plt.figure(figsize=(10, 5))
plt.plot(t, x, 'r', linewidth=2)
plt.title("Aproximación a onda cuadrada con armónicos impares", fontsize=14)
plt.xlabel("Tiempo t [s]", fontsize=12)
plt.ylabel("x(t)", fontsize=12)
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
w0 = 2 * np.pi   # frecuencia fundamental (rad/s)
T = 2 * np.pi / w0  # periodo (usando w0 = 2π => T = 1)
t = np.linspace(-1.5, 1.5, 3000)  # varios periodos para ver la repetición

# Función para construir suma de armónicos impares con signos alternados (serie tipo onda cuadrada)
def square_approx(t, w0, n_harmonics):
    """Aproxima una onda cuadrada usando n_harmonics impares (por ejemplo n_harmonics=5 usa 1,3,5,7,9)."""
    x = np.zeros_like(t)
    # tomar los primeros n_harmonics impares: 1,3,5,...
    count = 0
    n = 1
    while count < n_harmonics:
        signo = (-1) ** ((n - 1) // 2)
        x += signo * (1 / n) * np.cos(n * w0 * t)
        n += 2
        count += 1
    return x

# Señal original que diste (con DC 0.75 y armónicos hasta 9)
x_original = 0.75 + (2/3)*np.cos(w0*t) + (1/5)*np.cos(5*w0*t) - (1/7)*np.cos(7*w0*t) + (1/9)*np.cos(9*w0*t)

# Varias aproximaciones (sin DC) a la onda cuadrada
approxs = {
    "1 armónico (n=1)": square_approx(t, w0, 1),
    "3 armónicos (n=1,3,5)": square_approx(t, w0, 3),
    "9 armónicos (hasta n=17)": square_approx(t, w0, 9),
    "25 armónicos (hasta n=49)": square_approx(t, w0, 25)
}

# Onda cuadrada ideal (amplitud ±1, media 0)
square_ideal = np.sign(np.sin(w0 * t))  # devuelve -1, 0, 1; para evitar ceros exactos usar sign

# ----- Figura 1: Comparación de aproximaciones (sin DC) -----
plt.figure(figsize=(10, 4.5))
for label, sig in approxs.items():
    plt.plot(t, sig, label=label)
plt.plot(t, square_ideal, linewidth=1.0, label="Onda cuadrada ideal (±1)")
plt.xlim(-1, 1)
plt.title("Aproximaciones a una onda cuadrada (sin componente DC)")
plt.xlabel("t [s]")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend(loc="upper right")
plt.show()

# ----- Figura 2: Señal original (con DC) comparada con aproximación más rica (25 armónicos + DC) -----
plt.figure(figsize=(10, 4.5))
# aproximación rica con DC
approx_25_with_dc = 0.75 + square_approx(t, w0, 25)
plt.plot(t, x_original, linewidth=2.0, label="Señal original (hasta 9º)")
plt.plot(t, approx_25_with_dc, linewidth=1.0, label="Aproximación con 25 armónicos (más cercana a cuadrada) + DC=0.75")
# onda cuadrada ideal desplazada por DC=0.75 y amplitud ±1 -> tendrá rango [-1+0.75, 1+0.75] = [-0.25, 1.75]
plt.plot(t, 0.75 + square_ideal, linewidth=1.0, label="Onda cuadrada ideal desplazada (DC=0.75)")
plt.xlim(-1, 1)
plt.title("Comparación: señal original vs aproximación rica y onda cuadrada desplazada (DC=0.75)")
plt.xlabel("t [s]")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend(loc="upper right")
plt.show()

# ----- Figura 3: Zoom cerca de una discontinuidad para mostrar fenómeno de Gibbs -----
zoom_center = 0.5  # ubicamos un salto de la onda cuadrada en t = 0.5 (donde sinusoide cruza)
mask = (t > (zoom_center - 0.05)) & (t < (zoom_center + 0.05))
plt.figure(figsize=(10, 4.5))
plt.plot(t[mask], approx_25_with_dc[mask], linewidth=1.5, label="Aproximación con 25 armónicos + DC")
plt.plot(t[mask], x_original[mask], linewidth=2.0, label="Señal original (hasta 9º)")
plt.plot(t[mask], 0.75 + square_ideal[mask], linewidth=1.0, label="Onda cuadrada ideal desplazada")
plt.title("Zoom cerca de una discontinuidad (fenómeno de Gibbs)")
plt.xlabel("t [s]")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend(loc="upper right")
plt.show()
```
