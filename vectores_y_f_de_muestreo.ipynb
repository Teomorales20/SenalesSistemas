```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros de la recta
m = 2   # Pendiente
b = 1   # Intersección con el eje Y

# Rango de tiempo
t = np.linspace(-10, 10, 1000)  # desde -10 hasta 10, con 1000 puntos

# Definición de la señal
X_t = m * t + b

# Graficar la señal
plt.figure(figsize=(8, 4))
plt.plot(t, X_t, label=f'X(t) = {m}t + {b}', color='blue')
plt.axhline(0, color='black', linewidth=0.5)  # eje x
plt.axvline(0, color='black', linewidth=0.5)  # eje y
plt.title('Gráfica de la señal X(t) = mt + b')
plt.xlabel('Tiempo t')
plt.ylabel('Amplitud X(t)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Rango de tiempo
t = np.linspace(-10, 10, 1000)  # desde -10 hasta 10 con 1000 puntos

# Definición de la señal
X_t = t**2 + 5

# Graficar la señal
plt.figure(figsize=(8, 4))
plt.plot(t, X_t, label='X(t) = t² + 5', color='green')
plt.axhline(0, color='black', linewidth=0.5)  # eje x
plt.axvline(0, color='black', linewidth=0.5)  # eje y
plt.title('Gráfica de la señal X(t) = t² + 5')
plt.xlabel('Tiempo t')
plt.ylabel('Amplitud X(t)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
